```python
/Volumes/workspace/hdb/hdb-datasetimport pandas as pd
import glob

dbutils.widgets.text("master_data_file", "/Volumes/workspace/hdb/hdb-output-data/hdb_master_data.csv")
dbutils.widgets.text("input_folder", "/Volumes/workspace/hdb/hdb-dataset/*.csv")

master_data_file = dbutils.widgets.get("master_data_file")
input_folder = dbutils.widgets.get("input_folder")
```

```python
files = glob.glob(input_folder)
dfs = []
for file in files:
    dfs.append(pd.read_csv(file))

df = pd.concat(dfs, ignore_index=True)

df.to_csv(master_data_file, index=False)
```
